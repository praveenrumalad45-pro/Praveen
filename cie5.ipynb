```python
import spacy
from nltk import ngrams
nlp=spacy.load("en_core_web_sm")
doc=nlp("Ai simulation of human intelligence by machine it includes expects system nlp speech recognition and machine vision")
bigram=ngrams(doc,2)
print("bigram:",list(bigram))
trigram=ngrams(doc,3)
print("trigram:",list(trigram))
```
